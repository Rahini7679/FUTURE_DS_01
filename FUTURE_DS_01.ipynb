```python
import pandas as pd

df = pd.read_csv('Sample - Superstore.csv', encoding='latin1')
```

```python
df.isnull().sum()
```

```python
print("Total Sales =", df['Sales'].sum())
```

```python
print("Total Profit =", df['Profit'].sum())
```

```python
print("Total Orders =", df['Order ID'].nunique())
```

```python
total_quantity = df['Quantity'].sum()
print("Total Quantity Sold:", total_quantity)
```

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

print(category_sales)
```

```python
import matplotlib.pyplot as plt

category_sales.plot(kind='bar')

plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.savefig('sales_by_category.png', bbox_inches='tight')

plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

print(category_profit)
```

```python
category_profit.plot(kind='bar')

plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.savefig('profit_by_category.png', bbox_inches='tight')

plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

print(region_sales)
```

```python
region_sales.plot(kind='bar')

plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.savefig('sales_by_region.png', bbox_inches='tight')


plt.show()
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
top_products.plot(kind='bar')

plt.title('Top 10 Products by Sales')
plt.ylabel('Sales')
plt.savefig('top_10_products.png', bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Discount'], df['Profit'])

plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.savefig('discount_vs_profit.png', bbox_inches='tight')

plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
df['Month'] = df['Order Date'].dt.month
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()

monthly_sales.plot(marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.savefig('monthly_sales_trend.png', bbox_inches='tight')


plt.show()
```

```python
segment_sales = df.groupby('Segment')['Sales'].sum()

segment_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales by Customer Segment')
plt.ylabel('')
plt.savefig('sales_customer_segment.png', bbox_inches='tight')
plt.show()
```
